```python
df = pd.read_csv("car_sales_data.csv")
x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

print(df.head())
```

```python
df['Fuel type'].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6),
    title="Fuel type"
)
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Manufacturer", order=df["Manufacturer"].value_counts().index)
plt.title("Number of sales per car company ")
plt.xlabel("Manufacturer")
plt.ylabel(" Number of Sales")
```

```python
plt.figure(figsize=(8,11))
sns.scatterplot(data=df, x="Year of manufacture", y="Price")
plt.title("The relationship between the year of manufacture and the price of the car")
plt.xlabel("Model")
plt.ylabel("Price")
plt.show()

```

```python

plt.figure(figsize=(20,10))
sns.scatterplot(data=df, x="Mileage", y="Price")
plt.title("The relationship between the  Mileage and the price of the car")
plt.xlabel(" Mileage")
plt.ylabel("Price")
plt.show()
```

```python

sales_counts = df.groupby(['Manufacturer', 'Model']).size().reset_index(name='Count')

top3_sales = sales_counts.sort_values(by='Count', ascending=False).head(3)

print(top3_sales)
```
