```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("car_sales_data.csv")
x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

print(df.head())
```

```python
df['Fuel type'].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6),
    title="Fuel type",
)
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(8,11))
sns.scatterplot(data=df, x="Year of manufacture", y="Price")
plt.title("The relationship between the year of manufacture and the price of the car")
plt.xlabel("Model")
plt.ylabel("Price")
plt.show()

```

```python

plt.figure(figsize=(20,10))
sns.scatterplot(data=df, x="Mileage", y="Price")
plt.title("The relationship between the  Mileage and the price of the car")
plt.xlabel(" Mileage")
plt.ylabel("Price")
plt.show()
```
